```python
# Task 3: Calculator Program

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2

# Safe handling for division and modulus with zero
if num2 != 0:
    division = num1 / num2
    modulus = num1 % num2
else:
    division = "Undefined (division by zero)"
    modulus = "Undefined (division by zero)"

print("\n--- Calculator Results ---")
print(f"Addition ({num1} + {num2})       : {addition}")
print(f"Subtraction ({num1} - {num2})    : {subtraction}")
print(f"Multiplication ({num1} * {num2}) : {multiplication}")
print(f"Division ({num1} / {num2})       : {division}")
print(f"Modulus ({num1} % {num2})        : {modulus}")
```

```python
# Task 4: Grade Evaluation

marks = float(input("Enter your marks: "))

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "Fail"

print(f"Marks: {marks} -> Result: {grade}")
```

```python
# Task 5: Loops

# 1. Print numbers from 1 to 20 using a for loop
print("Numbers from 1 to 20:")
for i in range(1, 21):
    print(i, end=" ")
print("\n")

# 2. Print multiplication table of any number
table_num = int(input("Enter a number for the multiplication table: "))
print(f"Multiplication Table for {table_num}:")
for i in range(1, 11):
    print(f"{table_num} x {i} = {table_num * i}")
print()

# 3. Print even numbers from 1 to 50 using a while loop
print("Even numbers from 1 to 50:")
curr = 1
while curr <= 50:
    if curr % 2 == 0:
        print(curr, end=" ")
    curr += 1
print()
```

```python
# Task 9: Student Record Management System

students = []

def add_student():
    roll_no = input("Enter Roll Number: ").strip()
    # Check for unique roll number
    for s in students:
        if s["roll_no"] == roll_no:
            print("Error: A student with this Roll Number already exists!\n")
            return
    name = input("Enter Student Name: ").strip()
    branch = input("Enter Branch: ").strip()
    marks = input("Enter Marks: ").strip()

    student = {
        "roll_no": roll_no,
        "name": name,
        "branch": branch,
        "marks": marks
    }
    students.append(student)
    print(f"Student '{name}' added successfully!\n")

def display_students():
    if not students:
        print("No student records found.\n")
        return
    print("\n--- All Student Records ---")
    print(f"{'Roll No':<10} {'Name':<20} {'Branch':<15} {'Marks':<10}")
    print("-" * 55)
    for s in students:
        print(f"{s['roll_no']:<10} {s['name']:<20} {s['branch']:<15} {s['marks']:<10}")
    print()

def search_student():
    search_name = input("Enter student name to search: ").strip().lower()
    matches = [s for s in students if search_name in s["name"].lower()]

    if matches:
        print(f"\nFound {len(matches)} matching record(s):")
        for s in matches:
            print(f"Roll No: {s['roll_no']} | Name: {s['name']} | Branch: {s['branch']} | Marks: {s['marks']}")
    else:
        print(f"No student found with name matching '{search_name}'.")
    print()

def delete_student():
    roll_no = input("Enter Roll Number of the student to delete: ").strip()
    for s in students:
        if s["roll_no"] == roll_no:
            students.remove(s)
            print(f"Record for Roll No '{roll_no}' deleted successfully!\n")
            return
    print(f"No student found with Roll No '{roll_no}'.\n")

def main():
    while True:
        print("=== Student Record Management System ===")
        print("1. Add Student Details")
        print("2. Display All Student Records")
        print("3. Search Student by Name")
        print("4. Delete Student Record")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ").strip()
        print()

        if choice == "1":
            add_student()
        elif choice == "2":
            display_students()
        elif choice == "3":
            search_student()
        elif choice == "4":
            delete_student()
        elif choice == "5":
            print("Exiting Student Record Management System. Goodbye!")
            break
        else:
            print("Invalid selection. Please enter a choice between 1 and 5.\n")

if __name__ == "__main__":
    main()
```
