# Task 1: Statistical Analysis – Mean, Median & Mode


```python
mean_sales = df["Sales"].mean()

print("Mean Sales:", mean_sales)
```

```python
median_sales = df["Sales"].median()

print("Median Sales:", median_sales)
```

```python
mode_sales = df["Sales"].mode()

print("Mode Sales:", mode_sales.tolist())
```

# Task 2: Variance & Standard


```python
variance = df["Sales"].var()
std_deviation = df["Sales"].std()

print("Variance:", variance)
print("Standard Deviation:", std_deviation)
```

# Task 3: Correlation & Probability

```python
correlation = df["Sales"].corr(df["Profit"])

print("Correlation between Sales and Profit:", correlation)
```

```python
if correlation > 0:
    print("The relationship is positive.")
elif correlation < 0:
    print("The relationship is negative.")
else:
    print("There is no linear relationship.")
```

```python
technology_orders = (df["Category"] == "Technology").sum()
total_orders = len(df)

probability = technology_orders / total_orders

print("Technology Orders:", technology_orders)
print("Total Orders:", total_orders)
print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

# Task 4: Outlier Detection

```python
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[
    (df["Sales"] < lower_limit) |
    (df["Sales"] > upper_limit)
]

print("Number of Outliers:", len(outliers))
outliers[["Sales", "Profit", "Category"]].head(20)
```

```python
import pandas as pd

file_path = "/content/Sample - Superstore (1).xls"

df = pd.read_excel(file_path, sheet_name="Orders")

print("Columns:")
print(df.columns.tolist())

print("\nDataset Shape:")
print(df.shape)

df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

numeric_columns = ["Sales", "Quantity", "Discount", "Profit"]

correlation_matrix = df[numeric_columns].corr()

print("Correlation Matrix:")
print(correlation_matrix)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Superstore Dataset")
plt.show()
```

```python
print("Sales vs Profit Correlation:",
      df["Sales"].corr(df["Profit"]))

print("Sales vs Quantity Correlation:",
      df["Sales"].corr(df["Quantity"]))

print("Discount vs Profit Correlation:",
      df["Discount"].corr(df["Profit"]))
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print("Sales by Category:")
print(category_sales)
```

```python
category_profit = df.groupby("Category")["Profit"].sum()

print("Profit by Category:")
print(category_profit)
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

print("Sales by Region:")
print(region_sales)
```

```python
region_profit = df.groupby("Region")["Profit"].sum()

print("Profit by Region:")
print(region_profit)
```

```python
category_summary = df.groupby("Category").agg({
    "Sales": "sum",
    "Profit": "sum"
}).sort_values("Sales", ascending=False)

print(category_summary)
```

```python
region_summary = df.groupby("Region").agg({
    "Sales": "sum",
    "Profit": "sum"
}).sort_values("Profit", ascending=False)

print(region_summary)
```

```python
subcategory_summary = df.groupby("Sub-Category").agg({
    "Sales": "sum",
    "Profit": "sum"
}).sort_values("Profit", ascending=False)

print(subcategory_summary.head(10))
```
